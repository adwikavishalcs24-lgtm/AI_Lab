```
import random

def get_conflicts(state):
    """Calculate the number of conflicting pairs of queens."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Check if they are in the same row or diagonals
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_neighbors(state):
    """Generate all possible neighbor states by moving one queen in its column."""
    neighbors = []
    n = len(state)
    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbors.append(neighbor)
    return neighbors

def print_board(state):
    """Display the board visual representation."""
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

def hill_climbing_with_boards(n, max_restarts=100):
    """Solve the N-Queens problem and print every step's board and heuristic value."""
    for attempt in range(1, max_restarts + 1):
        print(f"\n{'='*15} Restart Attempt #{attempt} {'='*15}")
        # Initialize queens in random rows
        current_state = [random.randint(0, n - 1) for _ in range(n)]
        current_conflicts = get_conflicts(current_state)

        print(f"Initial random board state: {current_state} (Conflicts: {current_conflicts})")
        print_board(current_state)

        step = 0
        while True:
            if current_conflicts == 0:
                return current_state, attempt

            neighbors = get_neighbors(current_state)
            best_neighbor = None
            best_conflicts = current_conflicts

            # Evaluate neighbors
            for neighbor in neighbors:
                neighbor_conflicts = get_conflicts(neighbor)
                if neighbor_conflicts < best_conflicts:
                    best_conflicts = neighbor_conflicts
                    best_neighbor = neighbor

            if best_neighbor is None:
                print(f"\n[Stuck in local minimum/flat local maximum at step {step} with {current_conflicts} conflicts. Restarting...]\n")
                break

            step += 1
            current_state = best_neighbor
            current_conflicts = best_conflicts
            print(f"\nStep {step}: State updated to {current_state} (Conflicts: {current_conflicts})")
            print_board(current_state)

    return None, max_restarts

# User interaction
try:
    n_input = int(input("Enter the number of queens (N): "))
    if n_input < 4 and n_input != 1:
        print("No solutions exist for N < 4 (except N=1). Trying anyway...")

    print(f"Solving {n_input}-Queens with detailed board states...")
    solution, restarts = hill_climbing_with_boards(n_input)

    if solution:
        print(f"\n{'*'*15} Success! {'*'*15}")
        print(f"Solution found on restart attempt #{restarts}:")
        print(f"Final State representation: {solution}")
        print("\nFinal Board visualization:")
        print_board(solution)
    else:
        print(f"\nFailed to find a solution after {restarts} random restarts.")
except ValueError:
    print("Please enter a valid integer.")
```
