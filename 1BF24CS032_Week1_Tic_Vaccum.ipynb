```python
import random
import time

def print_board(board):
    """Prints the formatted 3x3 game board."""
    print("\n")
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---|---|---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---|---|---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print("\n")

def check_win(board, player):
    """Checks if the given player has won the game."""
    win_conditions = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]
    for condition in win_conditions:
        if all(board[i] == player for i in condition):
            return True
    return False

def check_draw(board):
    """Checks if the board is full without a winner."""
    return all(space in ['X', 'O'] for space in board)

def get_computer_move(board):
    """Simple AI: Wins if possible, blocks opponent win, or takes random space."""
    # 1. Check if computer can win in the next move
    for i in range(9):
        if board[i] not in ['X', 'O']:
            temp_board = board.copy()
            temp_board[i] = 'O'
            if check_win(temp_board, 'O'):
                return i

    # 2. Check if player can win in the next move and block them
    for i in range(9):
        if board[i] not in ['X', 'O']:
            temp_board = board.copy()
            temp_board[i] = 'X'
            if check_win(temp_board, 'X'):
                return i

    # 3. Take center if open
    if board[4] not in ['X', 'O']:
        return 4

    # 4. Take any open corner
    corners = [i for i in [0, 2, 6, 8] if board[i] not in ['X', 'O']]
    if corners:
        return random.choice(corners)

    # 5. Take any remaining available space
    available = [i for i in range(9) if board[i] not in ['X', 'O']]
    return random.choice(available)

def play_game():
    """Main game loop."""
    # Board initialized with position numbers (1-9) for easy selection
    board = [str(i + 1) for i in range(9)]
    human = 'X'
    computer = 'O'

    print("=" * 35)
    print(" Welcome to Tic-Tac-Toe in Colab! ")
    print("=" * 35)
    print("You are 'X' and the Computer is 'O'.")
    print("Select a space using numbers 1 to 9.")

    print_board(board)

    while True:
        # --- Human Turn ---
        while True:
            try:
                move = int(input("Your turn! Choose a position (1-9): ")) - 1
                if 0 <= move <= 8 and board[move] not in ['X', 'O']:
                    board[move] = human
                    break
                else:
                    print("Invalid move. Space is either occupied or out of range. Try again.")
            except ValueError:
                print("Please enter a valid number from 1 to 9.")

        print_board(board)

        if check_win(board, human):
            print("🎉 Congratulations! You beat the computer! 🎉")
            break

        if check_draw(board):
            print("🤝 It's a draw! Well played.")
            break

        # --- Computer Turn ---
        print("Computer is thinking...")
        time.sleep(0.8)
        comp_move = get_computer_move(board)
        board[comp_move] = computer

        print_board(board)

        if check_win(board, computer):
            print("🤖 Computer wins! Better luck next time.")
            break

        if check_draw(board):
            print("🤝 It's a draw! Well played.")
            break

# Run the game
if __name__ == "__main__":
    play_game()7
```

```python
import time

# ==========================================
# 1. SIMPLE REFLEX AGENT
# ==========================================
class SimpleReflexVacuumAgent:
    def __init__(self):
        pass

    def decide_action(self, location, status):
        """
        Condition-Action Rules (If-Then rules based ONLY on current percept)
        """
        if status == 'Dirty':
            return 'Suck'
        elif location == 'A':
            return 'Right'
        elif location == 'B':
            return 'Left'


# ==========================================
# 2. GOAL-BASED AGENT
# ==========================================
class GoalBasedVacuumAgent:
    def __init__(self):
        # Goal state: Both rooms must be Clean
        self.goal_state = {'A': 'Clean', 'B': 'Clean'}

        # Internal Model (Memory of the world state)
        self.world_model = {'A': 'Unknown', 'B': 'Unknown'}

    def decide_action(self, current_location, current_status):
        # 1. Update internal state/model based on current percept
        self.world_model[current_location] = current_status

        # 2. Check if the goal state is achieved
        if self.world_model['A'] == 'Clean' and self.world_model['B'] == 'Clean':
            return 'NoOp'  # Goal reached: stop working to save energy

        # 3. Plan action toward goal
        if current_status == 'Dirty':
            return 'Suck'
        elif current_location == 'A':
            return 'Right'
        elif current_location == 'B':
            return 'Left'


# ==========================================
# ENVIRONMENT SIMULATION
# ==========================================
class Environment:
    def __init__(self, initial_location='A', status_a='Dirty', status_b='Dirty'):
        self.agent_location = initial_location
        self.room_states = {'A': status_a, 'B': status_b}

    def get_percept(self):
        """Returns (location, status) of where the agent currently is."""
        return self.agent_location, self.room_states[self.agent_location]

    def execute_action(self, action):
        """Updates environment state based on agent's action."""
        if action == 'Suck':
            self.room_states[self.agent_location] = 'Clean'
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'
        # 'NoOp' does nothing


# ==========================================
# RUNNING THE COMPARISON
# ==========================================
def run_simulation(agent, env, agent_name, max_steps=6):
    print(f"\n==========================================")
    print(f"   Running Simulation: {agent_name}")
    print(f"==========================================")

    for step in range(1, max_steps + 1):
        location, status = env.get_percept()
        action = agent.decide_action(location, status)

        print(f"Step {step}: Location={location} | Status={status} | Action Taken: [{action}]")

        if action == 'NoOp':
            print(">> Goal achieved! Agent has shut down.")
            break

        env.execute_action(action)


# --- Execution ---
if __name__ == "__main__":
    # Test Environment Setup: Both rooms start DIRTY
    env_1 = Environment(initial_location='A', status_a='Dirty', status_b='Dirty')
    env_2 = Environment(initial_location='A', status_a='Dirty', status_b='Dirty')

    # Run Simple Reflex Agent
    reflex_agent = SimpleReflexVacuumAgent()
    run_simulation(reflex_agent, env_1, "Simple Reflex Agent", max_steps=6)

    # Run Goal-Based Agent
    goal_agent = GoalBasedVacuumAgent()
    run_simulation(goal_agent, env_2, "Goal-Based Agent", max_steps=6)
```
