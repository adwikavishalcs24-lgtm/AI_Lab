```
initial = (
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
)

goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)


def print_state(state, level):
    print(f"\nLevel {level}")
    for i in range(0, 9, 3):
        print(state[i:i+3])


def get_moves(state):
    moves = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    directions = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in directions:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col

            new_state = list(state)
            new_state[blank], new_state[new_blank] = (
                new_state[new_blank],
                new_state[blank]
            )

            moves.append((tuple(new_state), move))

    return moves


def depth_limited_dfs(state, depth, limit, path, visited):

    print_state(state, depth)

    if state == goal:
        print("\nGOAL FOUND!")
        return path

    if depth == limit:
        return None

    visited.add(state)

    for next_state, move in get_moves(state):

        if next_state not in visited:

            result = depth_limited_dfs(
                next_state,
                depth + 1,
                limit,
                path + [move],
                visited
            )

            if result is not None:
                return result

    return None


def ids():
    for limit in range(20):

        print("\n" + "=" * 30)
        print("DEPTH LIMIT =", limit)
        print("=" * 30)

        result = depth_limited_dfs(
            initial,
            0,
            limit,
            [],
            set()
        )

        if result is not None:
            print("\nSolution Path:")
            print(" → ".join(result))
            print("Solution Depth:", len(result))
            return


ids()
```
